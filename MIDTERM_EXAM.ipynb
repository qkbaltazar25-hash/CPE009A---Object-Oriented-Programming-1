```python
stock = input(str("Enter a combination of () or [] or {} where it is balanced (use only 3 times): "))

combinations = ["((()))", "[[[]]]", "{{{}}}", "{[()]}", "[{()}]", "{([])}", "{([])}", "([{}])", "[({})]"]

def checker():
  for i in range(len(combinations)):
    i += 1
    if stock == combinations[i-1]:
      print("True")
      break
    else:
      print("False")
      break



checker()
```
