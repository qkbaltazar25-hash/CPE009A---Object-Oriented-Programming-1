CLASS CONSTRUCTOR

```python
class Car:
  def __init__(self, name, mileage):
    self.name = name
    self.mileage = mileage
obj1 = Car("Toyota",2000)
obj2 = Car("Honda",3000)
print(obj1.name)
print(obj1.mileage)
print(obj2.name)
print(obj2.mileage)
```

```python
class Car:
  def __init__(self, name, mileage):
    self.name = name
    self.mileage = mileage
  def description(self):
    return f"The {self.name} car gives the mileage of {self.mileage}km/l"
car1 = Car("Toyota", 2000)
car2 = Car("Honda", 3000)
print(car1.description())
print(car2.description())
```

```python
class Animals:
  def __init__(self, name, habitat, temperature):
    self.name = name
    self.habitat = habitat
    self.temperature = temperature
  def feature(self):
    return f"The {self.name} which has a body temeperature {self.temperature} lives in {self.habitat}"
mammals = Animals("Dog", "Land", "Warm")
birds = Animals("Sparrow", "Air", "Warm")
reptiles = Animals("Snake", "Land", "Cold")
#print(mammals.name())
#print(mammals.habitat())
print(mammals.feature())
```

```python
import math
class Shapes:
  def __init__(self, name, sides, a = 0, b = 0):
    self.name = name
    self.sides = sides
    self.a = a
    self.b = b
  def method(self):
    if self.name == "Triangle":
      area = 0.5 * (self.a * self.b)
    elif self.name == "Circle":
      area = math.pi * (self.a * self.a)
    elif self.name == "Square":
      area = self.a * self.a
    elif self.name == "Rectangle":
      area = self.a * self.b
    return f"The {self.name} has {self.sides} sides and an area of {area}"
triangle = Shapes("Triangle", "3", 6, 7)
circle = Shapes("Circle", "0", 7)
square = Shapes("Square", "4", 5)
rectangle = Shapes("Rectangle", "4", 9, 11)

print(triangle.method())
print(circle.method())
print(square.method())
print(rectangle.method())
```
