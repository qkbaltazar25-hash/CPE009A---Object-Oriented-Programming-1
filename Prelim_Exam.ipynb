```python
print("Welcome to your grading calculator") #Welcome Statement
print("")
student_name = input("Enter the student name: ") #Student credentials
student_id = input("Enter the student ID: ")
#variables
course_counter = 0
choose = ""
totalmarks = 0
average = 0

while choose != "Y":
  course = input("Enter the course name/ course code: ") #Input course
  marks = int(input("Enter the marks: ")) #Input marks
  print("")

  if marks >= 90:
    grade = "A+"
    performace = "Outstanding"
  elif 89 >= marks >= 80:
    grade = "A"
    performace = "Excellent"
  elif 79 >= marks >= 70:
    grade = "B+"
    performace = "Very Good"
  elif 69 >= marks >= 60:
    grade = "B"
    performace = "Good"
  elif 59 >= marks >= 50:
    grade = "C"
    performace = "Satisfactory"
  elif 49 >= marks >= 40:
    grade = "D"
    performace = "Satisfactory"
  elif 39 >= marks >= 0:
    grade = "F"
    performace = "Needs Improvement"
  else:
    print("Invalid Input")

  print("Your grade is: ",grade) #prints A+ to F
  print("Your performane is: ",performace) #prints outstanding to needs improvement

  course_counter += 1  #adds 1 course within the system to divide the total marks and get the average
  totalmarks += marks
  average = (totalmarks / (course_counter))

  print("The average of all your courses is: ", average, "%") #prints the total average
  choose = (str(input("Do you Want to Exit? [Y|N]: "))) #verifies whether the user chooses to exit or not
  print("")
```
